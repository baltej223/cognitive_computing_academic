```python
# q1
roll = 1024170316
print(roll)

roll_arr = []
for i in range(0, 10):
    digit = int(roll%10)
    roll = int(roll/10)
    roll_arr.append(digit)

roll_arr.reverse()
# i)
print(roll_arr)
# ii)
roll_arr.append(6)
roll_arr.insert(7,4)

print(roll_arr)

# iii)
roll_arr.remove(4)
roll_arr.pop(5)
print(roll_arr)

# iv)
roll_arr.sort()
print(roll_arr)

roll_arr.sort(reverse=True)
print(roll_arr)

# v)
print(roll_arr[0:3])
print(roll_arr[7:10])

# vi)
av = sum(roll_arr)/len(roll_arr)
l = []
for i in range(0, 10):
    if roll_arr[i] > av:
        l.append(roll_arr[i])

print(l)
```
