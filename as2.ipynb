```python
# q1
roll = 1024170316
print(roll)

roll_arr = []
for i in range(0, 10):
    digit = int(roll%10)
    roll = int(roll/10)
    roll_arr.append(digit)

roll_arr.reverse()
# i)
print(roll_arr)
# ii)
roll_arr.append(6)
roll_arr.insert(7,4)

print(roll_arr)

# iii)
roll_arr.remove(4)
roll_arr.pop(5)
print(roll_arr)

# iv)
roll_arr.sort()
print(roll_arr)

roll_arr.sort(reverse=True)
print(roll_arr)

# v)
print(roll_arr[0:3])
print(roll_arr[7:10])

# vi)
av = sum(roll_arr)/len(roll_arr)
l = []
for i in range(0, 10):
    if roll_arr[i] > av:
        l.append(roll_arr[i])

print(l)
```

```python
# Q2

t = tuple(roll_arr[0:8])
print("Tuple:", t)

# i

m = max(t)
print("Index of Maximum element:", t.index(m))
print("Maximum element:", m)

# ii)
s = min(t)
print("Index of Minimum element:", t.index(s))
print("Minimum element:", s)
print("Frequency of Minimum element:", t.count(s))

# iii
rev = list(t[::-1])
print("Reversed tuple as list:", rev)

# iv)
score = int(input("Enter a score: "))
if score in t:
    print("First occurrence index:", t.index(score))
else:
    print("Score is not present in the tuple.")


# v)
try:
    t[0] = 100
except TypeError as e:
    print("Error:", e)
    print("Tuples are immutable, so their elements cannot be changed directly, unlike lists.")


# vi) 
first, second, *remaining = t
print("First score:", first)
print("Second score:", second)
print("Remaining scores:", remaining)
```

```python
# q3
import random
random.seed(1024170316)
arr = [random.randint(100, 900) for i in range(100)]

# i)
odd = [x for x in arr if x % 2 != 0]
print("Count of Odd numbers:", len(odd))

# ii)
even = [x for x in arr if x % 2 == 0]
print("Count of Even numbers:", len(even))


# iii)
prime = [x for x in arr if x > 1 and all(x % i != 0 for i in range(2, int(x ** 0.5) + 1))]
print("Count of Prime numbers:", len(prime))
print("Prime numbers:", prime)


# iv)
frequency = {}
for x in arr:
    frequency[x] = frequency.get(x, 0) + 1

m = max(frequency.values())
for x in frequency:
    if frequency[x] == m:
        print("Most Frequent number:", x)
        print("Frequency:", m)
        break
```
